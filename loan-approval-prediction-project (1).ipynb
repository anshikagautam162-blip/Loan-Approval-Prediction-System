```python
categorical_features = ['education', 'self_employed']

numerical_features = [
    'no_of_dependents',
    'income_annum',
    'loan_amount',
    'loan_term',
    'cibil_score',
    'residential_assets_value',
    'commercial_assets_value',
    'luxury_assets_value',
    'bank_asset_value',
    'loan_to_income_ratio'
]

print("Categorical:", categorical_features)
print("Numerical:", numerical_features)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_features)
    ]
)
```

```python
def predict_loan(
    no_of_dependents,
    education,
    self_employed,
    income_annum,
    loan_amount,
    loan_term,
    cibil_score,
    residential_assets_value,
    commercial_assets_value,
    luxury_assets_value,
    bank_asset_value
):
    
    # Calculate loan-to-income ratio
    loan_to_income_ratio = loan_amount / income_annum
    
    # Create input DataFrame
    input_data = pd.DataFrame({
        'no_of_dependents': [no_of_dependents],
        'education': [education],
        'self_employed': [self_employed],
        'income_annum': [income_annum],
        'loan_amount': [loan_amount],
        'loan_term': [loan_term],
        'cibil_score': [cibil_score],
        'residential_assets_value': [residential_assets_value],
        'commercial_assets_value': [commercial_assets_value],
        'luxury_assets_value': [luxury_assets_value],
        'bank_asset_value': [bank_asset_value],
        'loan_to_income_ratio': [loan_to_income_ratio]
    })
    
    # Apply the SAME preprocessing used during training
    input_processed = preprocessor.transform(input_data)
    
    # Make prediction
    prediction = best_tree.predict(input_processed)[0]
    
    # Get probabilities
    probabilities = best_tree.predict_proba(input_processed)[0]
    
    # Match probabilities with their class names
    class_probabilities = dict(
        zip(best_tree.classes_, probabilities)
    )
    
    return prediction, class_probabilities
```

```python
def loan_prediction_interface(
    no_of_dependents,
    education,
    self_employed,
    income_annum,
    loan_amount,
    loan_term,
    cibil_score,
    residential_assets_value,
    commercial_assets_value,
    luxury_assets_value,
    bank_asset_value
):

    # -----------------------------
    # Input validation
    # -----------------------------

    if not (0 <= no_of_dependents <= 5):
        return "⚠️ Invalid Input\n\nNumber of dependents must be between 0 and 5."

    if education not in ["Graduate", "Not Graduate"]:
        return "⚠️ Invalid Input\n\nPlease select a valid education category."

    if self_employed not in ["Yes", "No"]:
        return "⚠️ Invalid Input\n\nPlease select a valid self-employed option."

    if not (200000 <= income_annum <= 9900000):
        return (
            "⚠️ Invalid Input\n\n"
            "Annual income must be between ₹200,000 and ₹9,900,000 "
            "based on the training dataset."
        )

    if not (300000 <= loan_amount <= 39500000):
        return (
            "⚠️ Invalid Input\n\n"
            "Loan amount must be between ₹300,000 and ₹39,500,000 "
            "based on the training dataset."
        )

    if not (2 <= loan_term <= 20):
        return (
            "⚠️ Invalid Input\n\n"
            "Loan term must be between 2 and 20 years "
            "based on the training dataset."
        )

    if not (300 <= cibil_score <= 900):
        return (
            "⚠️ Invalid Input\n\n"
            "CIBIL score must be between 300 and 900."
        )

    if not (0 <= residential_assets_value <= 29100000):
        return (
            "⚠️ Invalid Input\n\n"
            "Residential asset value is outside the supported range."
        )

    if not (0 <= commercial_assets_value <= 19400000):
        return (
            "⚠️ Invalid Input\n\n"
            "Commercial asset value is outside the supported range."
        )

    if not (300000 <= luxury_assets_value <= 39200000):
        return (
            "⚠️ Invalid Input\n\n"
            "Luxury asset value must be between ₹300,000 and ₹39,200,000."
        )

    if not (0 <= bank_asset_value <= 14700000):
        return (
            "⚠️ Invalid Input\n\n"
            "Bank asset value is outside the supported range."
        )

    # -----------------------------
    # Prediction
    # -----------------------------

    # Removed overly restrictive loan_to_income_ratio check (was 1.5–4.0)
    # The model can handle any ratio; let it predict naturally
    loan_to_income_ratio = loan_amount / income_annum

    prediction, probabilities = predict_loan(
        no_of_dependents,
        education,
        self_employed,
        income_annum,
        loan_amount,
        loan_term,
        cibil_score,
        residential_assets_value,
        commercial_assets_value,
        luxury_assets_value,
        bank_asset_value
    )

    approved_prob = probabilities['Approved'] * 100
    rejected_prob = probabilities['Rejected'] * 100

    result = (
        "====================================\n"
        "       LOAN APPROVAL RESULT\n"
        "====================================\n\n"
        f"Loan Status: {prediction}\n\n"
        f"Approved Probability: {approved_prob:.2f}%\n"
        f"Rejected Probability: {rejected_prob:.2f}%\n\n"
        f"Loan-to-Income Ratio: {loan_to_income_ratio:.2f}"
    )

    return result
```

```python
# Convert the numerical Decision Tree thresholds
# back to their original real-world units.

scaler = preprocessor.named_transformers_['num']

numeric_features = [
    'no_of_dependents',
    'income_annum',
    'loan_amount',
    'loan_term',
    'cibil_score',
    'residential_assets_value',
    'commercial_assets_value',
    'luxury_assets_value',
    'bank_asset_value',
    'loan_to_income_ratio'
]

tree = best_tree.tree_

print("Original-unit Decision Tree thresholds:\n")

for node in range(tree.node_count):

    feature_index = tree.feature[node]

    if feature_index >= 0:

        feature_name = preprocessor.get_feature_names_out()[feature_index]
        threshold = tree.threshold[node]

        if feature_index < len(numeric_features):

            original_index = feature_index

            mean = scaler.mean_[original_index]
            std = scaler.scale_[original_index]

            original_threshold = threshold * std + mean

            print(
                f"Node {node}: "
                f"{feature_name} <= {original_threshold:.2f}"
            )

        else:
            print(
                f"Node {node}: "
                f"{feature_name} <= {threshold:.4f}"
            )
```
